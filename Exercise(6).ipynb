#### b) [10%] Basierend auf den test-Fällen, schreiben Sie eine Methode welche den Task erfolgreich (und effizient) löst.

Die Methode soll ebenfalls in eine separate Datei geschrieben werden.

```bash
"""
Modul zur Bestimmung des einzelnen Elements in einem Array,
in dem alle anderen Elemente genau zweimal vorkommen.
"""

from typing import List, Union


def find_single(arr: List[int]) -> int:
    """
    Findet das Element, das in einem Array genau einmal vorkommt,
    während alle anderen Elemente genau zweimal auftreten.

    Laufzeit: O(n), Speicher: O(1) – mittels XOR.

    Args:
        arr: Liste von ganzen Zahlen (nicht leer, ungerade Länge)

    Returns:
        Das einmal vorkommende Element.

    Raises:
        TypeError: falls arr keine Liste ist.
        ValueError: falls arr leer ist, oder die Bedingung nicht erfüllt ist.
    """
    if not isinstance(arr, list):
        raise TypeError("Eingabe muss eine Liste sein.")

    if len(arr) == 0:
        raise ValueError("Die Liste ist leer.")

    # XOR aller Elemente – doppelte Elemente heben sich auf
    result = 0
    for x in arr:
        result ^= x

    # Überprüfung, ob das gefundene Element tatsächlich einmal vorkommt
    count = 0
    for x in arr:
        if x == result:
            count += 1

    if count != 1:
        raise ValueError(
            "Die Liste enthält nicht genau ein Element, das genau einmal vorkommt."
        )

    return result
```
